```python
# importing the necessary libraries
from mlxtend.plotting import plot_decision_regions
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
sns.set()
import warnings
warnings.filterwarnings('ignore')
%matplotlib inline
```

Basic Data Science 

```python
#Loading the dataset
diabetes_data = pd.read_csv('diabetes.csv')
```

```python
#Print the first 5 rows of the dataframe.
diabetes_data.head()
```

Basic EDA and statistical analysis

```python
diabetes_data.info(verbose=True)
```

```python
diabetes_data.describe()
```

```python
diabetes_data.describe().T 
```

```python
diabetes_data_copy = diabetes_data.copy(deep=True)  # creating the copy of the dataset

# replacing the 0 values with NaN
diabetes_data_copy[['Glucose','BloodPressure','SkinThickness','Insulin','BMI']] = \
diabetes_data_copy[['Glucose','BloodPressure','SkinThickness','Insulin','BMI']].replace(0, np.nan)

# showing the count of NaNs
print(diabetes_data_copy.isnull().sum())
```

```python
p = diabetes_data_copy.hist(figsize = (20,20))
```

```python
diabetes_data_copy['Glucose'].fillna(diabetes_data_copy['Glucose'].mean(), inplace = True)
diabetes_data_copy.isna().sum()
```

```python
diabetes_data_copy['BloodPressure'].fillna(diabetes_data_copy['BloodPressure'].mean(), 
                                           inplace = True)
diabetes_data_copy.isna().sum()
```

```python
diabetes_data_copy['SkinThickness'].fillna(diabetes_data_copy['SkinThickness'].median(),
                                           inplace = True)
diabetes_data_copy.isna().sum()
```

```python
diabetes_data_copy['Insulin'].fillna(diabetes_data_copy['Insulin'].median(), inplace = True)
diabetes_data_copy.isna().sum()
```

```python
diabetes_data_copy['BMI'].fillna(diabetes_data_copy['BMI'].median(), inplace = True)
diabetes_data_copy.isna().sum()
```
